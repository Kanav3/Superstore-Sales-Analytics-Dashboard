```python
import pandas as pd
df=pd.read_csv('Sample - Superstore.csv', encoding='latin1')
df
```

### 1. Data Inspection & Quality Check
Checking column data types, missing values, and duplicate records across 9,994 transactions.

```python
print(df.info())
print('\nMissing values:\n',df.isnull().sum())
print('\nDuplicated rows:',df.duplicated().sum())
```

### 2. Datetime Parsing & Feature Engineering
Converting `Order Date` and `Ship Date` from string (`object`) to `datetime64`, and deriving `Order Year`, `Order Month`, `Shipping Days`, and `Profit Margin %`.

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month_name()
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Profit Margin %'] = round((df['Profit']/df['Sales']) * 100,2)
```

### 3. Export Cleaned Dataset for Power BI

```python
df.to_csv('cleaned_superstore.csv', index=False)
```
